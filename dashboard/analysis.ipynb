```python
import pandas as pd 
df = pd.read_csv(
    "../data/player_events.csv",
    parse_dates=["event_time"]
)
df.head()
```

```python
players = df["player_id"].nunique()

players
```

```python
day1_players = df.loc[df["day"] == 1, "player_id"].nunique()

d1_retention = day1_players/players
print(f"D1 Retention:{d1_retention:.2%}")
```

```python
for day in [1,3,7]:
    retention = df.loc[
        df["day"] == day,
        "player_id"].nunique()

    rate = retention/players
    print(f"D{day} Retention: {rate:.2%}")
    # print(f"D{day} Retention: {rate:.2%}")
```

```python
upgraded_player = set(
    df.loc[
        (df["day"] == 0)
        & (df["event_name"] =="hero_upgrade"),
        "player_id"
    ]
)

day1_players = set(
    df.loc[
        df["day"] == 1,
        "player_id"
    ]
)

# all_players = set(df["player_id"].nunique())
all_players = set(df["player_id"].unique())
```

```python
upgrade_retention = (len(upgraded_player & day1_players)/ 
                     len(upgraded_player))

non_upgrade_players = (all_players - upgraded_player )

non_upgrade_retention = (
    len(non_upgrade_players & day1_players) / len(non_upgrade_players)
)

print(
    f"Upgraded D1 retention: "
    f"{upgrade_retention:.2%}"
)

print(
    f"No upgrade D1 retention: "
    f"{non_upgrade_retention:.2%}"
)
```

Players completing their first hero upgrade during Day 0 show higher D1 retention than players who do not
